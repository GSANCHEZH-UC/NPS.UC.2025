## Aplicando SMOTE antes de entrenar el MLPClassifier

Este proceso implica los siguientes pasos:
1.  Dividir los datos en conjuntos de entrenamiento y prueba.
2.  Aplicar SMOTE **solo** al conjunto de entrenamiento para generar muestras sintéticas de la clase minoritaria.
3.  Entrenar el `MLPClassifier` con el conjunto de entrenamiento balanceado.
4.  Evaluar el modelo con el conjunto de prueba original (sin SMOTE).


```python
from imblearn.over_sampling import SMOTE
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Asumiendo que X y y ya están definidos del análisis anterior
# X = Promotores[['SATIF_ACAD', 'SATISF_SERV', 'SATISF_EXP', 'SATISF_REG_ACAD', 'SATISF_ACT_SOST', 'SATISF_APY_LOGRO_ACAD', 'SATISF_INF_TEC', 'SATISF_CAMP_VIRT', 'SATISF_COM_IMAG']]
# y = Promotores['PROMOTOR']

# 1. Dividir los datos en conjuntos de entrenamiento y prueba
X_train_smote, X_test_smote, y_train_smote, y_test_smote = train_test_split(XCopia, y, test_size=0.3, random_state=42)

# 2. Aplicar SMOTE al conjunto de entrenamiento
sm = SMOTE(random_state=42)
X_train_res, y_train_res = sm.fit_resample(X_train_smote, y_train_smote)

print(f'Shape of X_train before SMOTE: {X_train_smote.shape}')
print(f'Shape of X_train after SMOTE: {X_train_res.shape}')
print(f'Class distribution of y_train before SMOTE:\n{y_train_smote.value_counts()}')
print(f'Class distribution of y_train after SMOTE:\n{y_train_res.value_counts()}')
```

### Entrenar MLPClassifier con datos balanceados por SMOTE

```python
# 3. Entrenar el MLPClassifier con el conjunto de entrenamiento balanceado
# Se usa el mejor modelo MLP encontrado anteriormente, pero sin class_weight
# ya que SMOTE ya ha balanceado las clases.
mlp_smote = MLPClassifier(
    hidden_layer_sizes=(50,),
    activation='relu',
    solver='adam',
    alpha=0.001,
    learning_rate_init=0.001,
    max_iter=200,
    random_state=50
)

mlp_smote.fit(X_train_res, y_train_res)

# 4. Evaluar el modelo con el conjunto de prueba original
y_pred_mlp_smote = mlp_smote.predict(X_test_smote)

accuracy_mlp_smote = accuracy_score(y_test_smote, y_pred_mlp_smote)
report_mlp_smote = classification_report(y_test_smote, y_pred_mlp_smote)

print(f'Precisión del modelo MLP con SMOTE: {accuracy_mlp_smote:.3f}')
print(f'Reporte de clasificación del modelo MLP con SMOTE:\n{report_mlp_smote}')
```
